```python
import torch
import matplotlib.pyplot as plt
import torchvision.utils as vutils

# Load
run_name = "bumbling-thunder-13"  # replace with your actual run name
data = torch.load(f"logged_files/distillation/waterbirds/dinov2_vitb/{run_name}/data.pth", weights_only=False)
print(data.keys())  # voir ce qu'il y a dedans

images = data["images"]  # (N, C, H, W)
print(f"Shape: {images.shape}")

# Normalize to [0, 1] for display
def normalize(img):
    img = img - img.min()
    img = img / img.max()
    return img

# Display grid
grid = vutils.make_grid(normalize(images), nrow=4, padding=2)
plt.figure(figsize=(12, 12))
plt.imshow(grid.permute(1, 2, 0).cpu().numpy())
plt.axis("off")
plt.title("Distilled Images")
plt.show()
print("Saved to distilled.png")
```

```python
from data.dataloaders.fish import Fish4Knowledge
import torch
import matplotlib.pyplot as plt
import torchvision.utils as vutils

train = Fish4Knowledge(split="train")

images = []
for c in range(train.num_classes):
    first_local_idx = (train.targets == c).nonzero(as_tuple=True)[0][0].item()
    img, _ = train[first_local_idx]
    images.append(img)
    print(img.shape)
grid = vutils.make_grid(torch.stack(images), nrow=4, padding=2)
plt.figure(figsize=(12, 12))
plt.imshow(grid.permute(1, 2, 0).cpu().numpy())
plt.axis("off")
plt.title("Exemples d'images de chaque classe du dataset Fish4Knowledge")
plt.show()
```

```python
import os
import matplotlib.pyplot as plt

base_path = "/Users/alex/Developpement/Internship/fish4konwledge/fish_image"

classes = sorted([d for d in os.listdir(base_path) if os.path.isdir(os.path.join(base_path, d))])
counts = [len(os.listdir(os.path.join(base_path, c))) for c in classes]

plt.figure(figsize=(14, 5))
bars = plt.bar(classes, counts, color="steelblue", edgecolor="black", linewidth=0.5)

# Valeur au dessus de chaque barre
for bar, count in zip(bars, counts):
    plt.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + 50,
             str(count), ha="center", va="bottom", fontsize=8)

plt.xticks(rotation=45, ha="right", fontsize=9)
plt.ylabel("Number of images")
plt.title("Fish4Knowledge — Class Distribution")
plt.tight_layout()
plt.show()
```
